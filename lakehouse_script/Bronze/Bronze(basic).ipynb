##Reading from csv file and Writing to the Delta table


```python
df=(
    spark.read.option("header","true").option ("inferSchema","true").csv("/Volumes/workspace/bronze/source_systems/cust_info.csv")
)
df.write.mode("overwrite").saveAsTable("workspace.bronze.crm_cust_info")
```
